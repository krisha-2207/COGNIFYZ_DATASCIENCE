```python
# ==========================
# LEVEL 2 - TASK 1
# Table Booking & Online Delivery
# ==========================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Dataset .csv")

# --------------------------
# Percentage of Table Booking
# --------------------------

table_booking_percent = df["Has Table booking"].value_counts(normalize=True) * 100

print("Percentage of Restaurants with Table Booking")
print(table_booking_percent)

# --------------------------
# Percentage of Online Delivery
# --------------------------

online_delivery_percent = df["Has Online delivery"].value_counts(normalize=True) * 100

print("\nPercentage of Restaurants with Online Delivery")
print(online_delivery_percent)

# --------------------------
# Average Rating Comparison
# --------------------------

avg_rating = df.groupby("Has Table booking")["Aggregate rating"].mean()

print("\nAverage Rating based on Table Booking")
print(avg_rating)

# --------------------------
# Online Delivery by Price Range
# --------------------------

delivery_price = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"],
    normalize="index"
) * 100

print("\nOnline Delivery Percentage by Price Range")
print(delivery_price)

# --------------------------
# Visualization 1
# --------------------------

plt.figure(figsize=(6,5))
sns.countplot(x="Has Table booking", data=df)
plt.title("Restaurants Offering Table Booking")
plt.xlabel("Table Booking")
plt.ylabel("Number of Restaurants")
plt.show()

# --------------------------
# Visualization 2
# --------------------------

plt.figure(figsize=(6,5))
sns.countplot(x="Has Online delivery", data=df)
plt.title("Restaurants Offering Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")
plt.show()

# --------------------------
# Visualization 3
# --------------------------

plt.figure(figsize=(6,5))
sns.barplot(
    x=avg_rating.index,
    y=avg_rating.values
)

plt.title("Average Rating by Table Booking")
plt.xlabel("Table Booking")
plt.ylabel("Average Rating")
plt.show()

# --------------------------
# Visualization 4
# --------------------------

delivery_price.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Online Delivery Availability Across Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Percentage")
plt.legend(title="Online Delivery")
plt.show()

print("\n Level 2 - Task 1 Completed Successfully!")
```

```python
# =====================================
# LEVEL 2 - TASK 2
# Price Range Analysis
# =====================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Dataset .csv")

# -------------------------------
# Most Common Price Range
# -------------------------------

print("Most Common Price Range:")
print(df["Price range"].value_counts())

most_common = df["Price range"].mode()[0]
print(f"\nMost Common Price Range is: {most_common}")

# -------------------------------
# Average Rating for each Price Range
# -------------------------------

price_rating = df.groupby("Price range")["Aggregate rating"].mean()

print("\nAverage Rating by Price Range:")
print(price_rating)

# -------------------------------
# Rating Color Analysis
# -------------------------------

color_rating = df.groupby("Rating color")["Aggregate rating"].mean().sort_values(ascending=False)

print("\nAverage Rating by Rating Color:")
print(color_rating)

highest_color = color_rating.idxmax()

print(f"\nRating Color with Highest Average Rating: {highest_color}")

# -------------------------------
# Visualization 1
# -------------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    x="Price range",
    data=df
)

plt.title("Distribution of Restaurants by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()

# -------------------------------
# Visualization 2
# -------------------------------

plt.figure(figsize=(7,5))

sns.barplot(
    x=price_rating.index,
    y=price_rating.values
)

plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")
plt.show()

# -------------------------------
# Visualization 3
# -------------------------------

plt.figure(figsize=(8,5))

sns.barplot(
    x=color_rating.index,
    y=color_rating.values
)

plt.title("Average Rating by Rating Color")
plt.xlabel("Rating Color")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)

plt.show()

print("\nLevel 2 - Task 2 Completed Successfully!")
```

```python
# =====================================
# LEVEL 2 - TASK 3
# Feature Engineering
# =====================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Dataset .csv")

# -----------------------------------
# Restaurant Name Length
# -----------------------------------

df["Restaurant Name Length"] = df["Restaurant Name"].str.len()

# -----------------------------------
# Address Length
# -----------------------------------

df["Address Length"] = df["Address"].str.len()

# -----------------------------------
# Encode Table Booking
# -----------------------------------

df["Table Booking"] = df["Has Table booking"].map({
    "Yes":1,
    "No":0
})

# -----------------------------------
# Encode Online Delivery
# -----------------------------------

df["Online Delivery"] = df["Has Online delivery"].map({
    "Yes":1,
    "No":0
})

# -----------------------------------
# Display New Features
# -----------------------------------

print(df[[
    "Restaurant Name",
    "Restaurant Name Length",
    "Address",
    "Address Length",
    "Table Booking",
    "Online Delivery"
]].head())

# -----------------------------------
# Statistics
# -----------------------------------

print("\nFeature Statistics")
print(df[[
    "Restaurant Name Length",
    "Address Length",
    "Table Booking",
    "Online Delivery"
]].describe())

# -----------------------------------
# Correlation
# -----------------------------------

corr = df[[
    "Restaurant Name Length",
    "Address Length",
    "Table Booking",
    "Online Delivery",
    "Aggregate rating"
]].corr()

print("\nCorrelation Matrix")
print(corr)

# -----------------------------------
# Heatmap
# -----------------------------------

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Engineered Features")

plt.show()

# -----------------------------------
# Distribution of Restaurant Name Length
# -----------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["Restaurant Name Length"],
    bins=20,
    kde=True
)

plt.title("Distribution of Restaurant Name Length")

plt.show()

# -----------------------------------
# Distribution of Address Length
# -----------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["Address Length"],
    bins=20,
    kde=True
)

plt.title("Distribution of Address Length")

plt.show()

print("\nLevel 2 - Task 3 Completed Successfully!")
```
